```python
import sqlite3
import random
import pandas as pd
```

```python
# Create connection
conn = sqlite3.connect("employees.db")
cursor = conn.cursor()

# Create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS employees (
    emp_id INTEGER PRIMARY KEY,
    name TEXT,
    department TEXT,
    salary REAL,
    years_experience INTEGER,
    performance_score REAL
)
""")

departments = ["Engineering", "Sales", "Marketing", "HR", "Finance"]

# Generate 40 employee records
employees = []

for i in range(1, 41):
    name = f"Employee_{i}"
    department = random.choice(departments)
    salary = random.randint(50000,150000)
    years_experience = random.randint(1,15)
    performance_score = round(random.uniform(1.0,5.0),2)

    employees.append((i,name,department,salary,years_experience,performance_score))

# Insert data
cursor.executemany("""
INSERT INTO employees
(emp_id,name,department,salary,years_experience,performance_score)
VALUES (?,?,?,?,?,?)
""", employees)

conn.commit()

print("Database created with 40 employee records.")
```

```python
query1 = """
SELECT name, department, salary, performance_score
FROM employees
WHERE performance_score >= 4.0
AND years_experience >= 3
ORDER BY performance_score DESC
LIMIT 15
"""

df_q1 = pd.read_sql_query(query1, conn)
df_q1
```

```python
query2 = """
SELECT *
FROM employees
WHERE salary BETWEEN 70000 AND 110000
AND department IN ('Engineering','Sales')
ORDER BY department, salary DESC
"""

df_q2 = pd.read_sql_query(query2, conn)
df_q2
```

```python
query3 = """
SELECT department,
COUNT(*) AS employee_count,
AVG(salary) AS avg_salary
FROM employees
GROUP BY department
ORDER BY avg_salary DESC
"""

df_q3 = pd.read_sql_query(query3, conn)
df_q3
```

```python
df_all = pd.read_sql_query("SELECT * FROM employees", conn)

avg_perf = df_all.groupby("department")["performance_score"].mean()

avg_perf
```

```python
pandas_q1 = (
    df_all
    .loc[(df_all["performance_score"] >= 4.0) & (df_all["years_experience"] >= 3),
         ["name","department","salary","performance_score"]]
    .sort_values(by="performance_score", ascending=False)
    .head(15)
)

pandas_q1
```

## SQL vs Pandas Comparison

SQL and Pandas are both powerful tools for data analysis but they differ in syntax and typical usage. SQL uses expressions such as `=` for equality, `AND` for combining conditions, and `ORDER BY` for sorting results. In contrast, Pandas uses Python-style syntax such as `==` for equality, `&` for logical AND, and `.sort_values()` for sorting data. SQL queries are typically executed directly on databases, while Pandas operations are performed on DataFrames loaded into memory.

SQL is particularly useful when working with relational databases and structured data stored on disk. It allows analysts to filter, aggregate, and join large datasets efficiently without loading the entire dataset into memory. Pandas, on the other hand, is ideal for in-memory data manipulation, exploratory data analysis, and integration with Python's data science ecosystem such as NumPy, Matplotlib, and Scikit-learn.

For datasets that are too large to fit into memory, SQL is generally the better choice because database engines are optimized for disk-based storage, indexing, and query optimization. Pandas performs best when datasets comfortably fit into RAM and when complex transformations or statistical analysis are required.
